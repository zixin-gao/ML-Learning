```python
import torch
```

```python
x = torch.arange(12)
x
```

```python
x.shape
```

```python
x.numel()
```

```python
X = x.reshape(3,4)
```

```python
X
```

```python
X[0:2,:] = 12
X
```

```python
Z = torch.zeros_like(X)
Z
print(f'id(Z): {id(Z)}')
```

```python
Z[:] = X + X  # gives the same address else address will change
print(f'id(Z): {id(Z)}')
```

```python
Z += X
print(f'id(Z): {id(Z)}')
```
