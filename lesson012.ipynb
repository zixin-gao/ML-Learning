```python
import torch

x = torch.arange(4.0)
x
```

```python
x.requires_grad_(True)
x.grad  # this is where the gradients of y wrt x is stored, defaults to None
```

```python
y = 2 * torch.dot(x,x)
y
```

```python
y.backward()
x.grad
```

```python
x.grad == 4 * x
```

```python
x.grad.zero_()
y = x.sum()
y.backward()
x.grad
```

```python
# if its not a vector, then take the sum of that vector before taking the gradient
x.grad.zero_()
y = x*x
y.sum().backward()
x.grad
```

```python
x.grad.zero_()
y = x*x
u = y.detach()
z = u * x
z.sum().backward()
x.grad==u
```

```python
def f(a):
    b = a * 2
    while b.norm() < 1000:
        b = b * 2
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    return c

a = torch.randn(size=(), requires_grad = True)
print(a)
d = f(a)
print(d)
d.backward()

a.grad == d/a  # because anything in the function prduces a linear relationship between d and a
```
