```python
import numpy as np
import torch
from torch.utils import data

import os
import sys
sys.path.insert(0, './d2l-en/pytorch')
from d2l import torch as d2l
```

```python
true_w = torch.tensor([2, -3.4])
true_b = 4.2
features, labels = d2l.synthetic_data(true_w, true_b, 1000)
```

```python
def load_array(data_arrays, batch_size, is_train=True):
    dataset = data.TensorDataset(*data_arrays)
    # * is an unpacking operator into individual elements
    # because data_array in the form of (features, labels), and TensorDataset takes features, labels separately, not as a tuple
    # TensorDataset pairs features with labels
    return data.DataLoader(dataset, batch_size, shuffle=is_train)

batch_size = 10
data_iter = load_array((features, labels), batch_size)

next(iter(data_iter))
# iter is a class
# next is a function that returns the next item
```

```python
from torch import nn
# nn for neuraal network

net = nn.Sequential(nn.Linear(2,1)) 
# only one layer = linear regression
# sequential puts all layers in order
# Linear(2,1) means 2 input layers, 1 output layer
```

```python
# initialize model parameters

# by default, PyTorch initializes weights randomly
net[0].weight.data.normal_(0, 0.01)
# net[0] access first layer in sequential
# .weight access weight parameter
# .data access the actual tensor
# .normal_(0, 0.01) fill with values from normal distribution

net[0].bias.data.fill_(0)
# bias set all to just 0
```

```python
# get loss from the loss function
loss = nn.MSELoss()
```

```python
trainer = torch.optim.SGD(net.parameters(), lr = 0.03)
# SGD = stochastic gradient descent
# implements: parameter = parameter - learning_rate * gradient
# net.parameters: returns all trainable parameters of neural network
```

```python
num_epochs = 3
for epoch in range(num_epochs):
    for X, y in data_iter:
        l = loss(net(X), y)
        trainer.zero_grad()  # clear all the previous gradients
        l.backward()
        trainer.step()
    l = loss(net(features), labels)
    print(f'epoch {epoch + 1}, loss {l:f}')
```
