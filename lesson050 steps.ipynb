```python
import torch
from torch import nn
def comp_conv2d(conv2d, X):
    X = X.reshape((1, 1,) + X.shape)
    # conv2d expects a 4D input of shape (batch_size, in_channels, height, width)
    # reshape X to have a batch size of 1 and 1 input channel
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])
    # reshape the output back to 2D by removing the batch and channel dimensions
    # the output shape is (height, width)

conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1)
X = torch.rand(size=(8, 8))
comp_conv2d(conv2d, X).shape
```

filling with different height and width

```python
conv2d = nn.Conv2d(1, 1, kernel_size=(5, 3), padding=(2, 1))
# The kernel size is (5, 3) and the padding is (2, 1) to maintain the same height and width
# padding (2, 1) means we add 2 rows of 0s to the top and bottom, and 1 column of 0s to the left and right
comp_conv2d(conv2d, X).shape
```

stride

```python
conv2d = nn.Conv2d(1, 1, kernel_size=3, padding=1, stride=2)
comp_conv2d(conv2d, X).shape
```

```python
conv2d = nn.Conv2d(1, 1, kernel_size=(3, 5), padding=(0, 1), stride=(3, 4))
comp_conv2d(conv2d, X).shape
```
