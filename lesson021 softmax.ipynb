```python
import torch
import os
import sys
from torch import nn
sys.path.insert(0, './d2l-en/pytorch')
from d2l import torch as d2l
```

```python
batch_size = 256
train_iter, test_iter = d2l.load_data_fashion_mnist(batch_size)
```

```python
net = nn.Sequential(nn.Flatten(), nn.Linear(784, 10))
# first layer flattens input data into 2d array
# second layer takes the 784 pixels and put 10 outputs (1 for each type of image)

def init_weights(m):
    # check if module m is a linear layer
    # if yes, initialize its weights with normal distribution
    if type(m) == nn.Linear:
        nn.init.normal_(m.weight, std=0.01)

net.apply(init_weights);
# .apply() recursively applies a function to every module in net
# it calls init_weights on
    # 1. the flatten layer -> skipped because not linear
    # 2. the linear layer (weights initialized) 
```

```python
loss = nn.CrossEntropyLoss()
```
